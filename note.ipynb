1. `belongsTo` => Relasi yang menghubungkan dua tabel melalui foreign key dimana model satu relasinya berada di sisi yang mempunyai foreign key. 
<br>`Contoh Kasus` : 
Tabel `comments` memiliki foreign key `user_id` yang merujuk pada tabel `users`. Ini menunjukkan bahwa setiap komentar dibuat oleh satu pengguna tertentu. Jadi, relasi antara tabel comments dan users adalah belongsTo, karena setiap komentar milik satu pengguna.

```php
class Comment extends Model
{
    public function user()
    {
        return $this->belongsTo(User::class);
    }
}
```

2. `belongsToMany` : Relasi banyak ke banyak antara dua tabel dengan menggunakan tabel pivot yang berisi foreign key dari masing-masing tabel.
<br>`Contoh Kasus`  : Tabel `users` memiliki relasi banyak ke banyak dengan tabel `roles` dan menggunakan tabel `pivot role_user`. Ini menunjukkan bahwa seorang pengguna dapat memiliki banyak peran, dan setiap peran dapat dimiliki oleh banyak pengguna

```php
class User extends Model
{
    public function roles()
    {
        return $this->belongsToMany(Role::class);
    }
}
```

3. `hasOne` : Relasi yang menghubungkan dua tabel melalui foreign key dimana model satu relasinya berada di sisi yang memiliki primary key.
<br>`Contoh Kasus` : Tabel `users` memiliki kolom `profile_id` yang merujuk pada tabel `profiles`. Ini menunjukkan bahwa setiap pengguna memiliki satu profil. Jadi, relasi antara tabel users dan profiles adalah hasOne, karena setiap pengguna memiliki satu profil.

```php
class User extends Model
{
    public function profile()
    {
        return $this->hasOne(Profile::class);
    }
}
```

Berikut adalah penjelasan dan contoh penggunaan beberapa parameter pada relasi Laravel:

foreign_key: Parameter ini digunakan untuk menentukan nama field kunci asing yang digunakan dalam relasi. Secara default, Laravel akan mencari field kunci asing dengan format nama_tabel + _id, misalnya user_id untuk relasi antara tabel users dan posts. Namun, jika nama field kunci asing Anda berbeda, maka Anda dapat menentukan parameter ini untuk menyesuaikannya.
Contoh:

kotlin
Copy code
class Post extends Model
{
    public function user()
    {
        return $this->belongsTo(User::class, 'author_id');
    }
}
Pada contoh di atas, field kunci asing author_id digunakan sebagai kunci asing pada relasi Post dengan User.

local_key: Parameter ini digunakan untuk menentukan nama field kunci primer pada tabel model saat ini. Secara default, Laravel akan mencari field kunci primer dengan nama id, namun Anda dapat menentukan parameter ini jika nama field kunci primer Anda berbeda.
Contoh:

kotlin
Copy code
class Post extends Model
{
    public function user()
    {
        return $this->belongsTo(User::class, 'author_id', 'user_id');
    }
}
Pada contoh di atas, field kunci primer user_id pada tabel users digunakan untuk mencocokkan field kunci asing author_id pada tabel posts.

table: Parameter ini digunakan untuk menentukan nama tabel pivot pada relasi many-to-many. Secara default, Laravel akan menggunakan format nama_tabel_model_1 + _ + nama_tabel_model_2 dengan urutan abjad untuk menentukan nama tabel pivot. Namun, jika nama tabel pivot Anda berbeda, maka Anda dapat menentukan parameter ini untuk menyesuaikannya.
Contoh:

kotlin
Copy code
class User extends Model
{
    public function roles()
    {
        return $this->belongsToMany(Role::class, 'user_role', 'user_id', 'role_id');
    }
}
Pada contoh di atas, relasi many-to-many antara User dan Role menggunakan tabel pivot dengan nama user_role.

through: Parameter ini digunakan pada relasi many-to-many untuk menentukan model "perantara" antara dua model yang sedang dihubungkan. Contohnya, jika Anda memiliki tabel orders, products, dan customers, dan ingin membuat relasi many-to-many antara Order dan Product melalui Customer, maka Anda dapat menggunakan parameter through untuk menentukan model Customer sebagai "perantara" antara Order dan Product.
Contoh:

kotlin
Copy code
class Order extends Model
{
    public function products()
    {
        return $this->belongsToMany(Product::class, Customer::class, 'order_id', 'product_id', 'customer_id', 'id');
    }
}
Pada contoh di atas, relasi many-to-many antara Order dan Product menggunakan model Customer sebagai "perantara". Kunci asing order_id dan product_id digunakan untuk menghubungkan Order dan Product ke Customer.
